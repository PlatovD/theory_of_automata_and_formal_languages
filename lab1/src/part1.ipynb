# Тестирую правильность составленного ДКА

```python
from writer import Writer
%load_ext autoreload
%autoreload 2

from reader import Reader, Descriptor
from runner import Runner, RunResult
from word_generator import WordGenerator

desc: Descriptor = Reader.read_from_file("./PlatovDK_lab1_part1.txt")
all_words_length_less_16 = WordGenerator.generate_words(desc.alphabet, 15)

for word in all_words_length_less_16:
    cnt_1_multiple_3 = (word.count("1") % 3) == 0
    ends_10 = word.endswith("10")

    success_word = ends_10 and not cnt_1_multiple_3

    res: RunResult = Runner.run_word(desc, word)
    if success_word != res.is_success:
        raise ValueError("Разные значения!")

Writer.write_to_file('res', desc)
        
```
